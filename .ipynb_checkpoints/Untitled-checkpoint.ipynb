```python
df_clean["Attendance_Risk"] = (
    df_clean["Attendance"] < 70
)
```

```python
X = df_clean.drop(
    columns=[
        "Exam_Score",
        "Academic_Risk",
        "Attendance_Risk"
    ]
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.value_counts(normalize=True)
```

```python
y_test.value_counts(normalize=True)
```

```python
numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = X.select_dtypes(
    include=["object"]
).columns
```

```python
print("Numerical columns:")
print(numerical_columns)
```

```python
print("Categorical columns:")
print(categorical_columns)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(handle_unknown="ignore")
    )
])
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    (
        "numerical",
        numerical_pipeline,
        numerical_columns
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_columns
    )
])
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\nTraining target:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True))

print("\nTesting target:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.base import clone

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve
)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
model_estimators = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),

    "SVM": SVC(
        probability=True,
        random_state=42
    )
}
```

```python
results = []
fitted_models = {}

for model_name, estimator in model_estimators.items():

    pipeline = Pipeline([
        ("preprocessor", clone(preprocessor)),
        ("model", clone(estimator))
    ])

    # Train model
    pipeline.fit(X_train, y_train)

    # Make predictions
    y_pred = pipeline.predict(X_test)

    # Get probability of Academic_Risk = 1
    y_prob = pipeline.predict_proba(X_test)[:, 1]

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )
    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )
    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )
    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "ROC_AUC": roc_auc
    })

    fitted_models[model_name] = pipeline
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by=["Recall", "F1", "ROC_AUC"],
    ascending=False
).reset_index(drop=True)

results_df
```

```python
best_model_name = results_df.iloc[0]["Model"]

print("Selected model:", best_model_name)

best_model = fitted_models[best_model_name]
```

```python
best_predictions = best_model.predict(X_test)

best_probabilities = best_model.predict_proba(
    X_test
)[:, 1]

print(
    classification_report(
        y_test,
        best_predictions,
        target_names=[
            "Not At Risk",
            "At Risk"
        ]
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    best_predictions
)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Not At Risk",
        "At Risk"
    ]
)

display.plot()

plt.title(
    f"Confusion Matrix - {best_model_name}"
)

plt.show()
```

```python
plt.figure(figsize=(8, 6))

for model_name, model in fitted_models.items():

    probabilities = model.predict_proba(
        X_test
    )[:, 1]

    fpr, tpr, thresholds = roc_curve(
        y_test,
        probabilities
    )

    auc_score = roc_auc_score(
        y_test,
        probabilities
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{model_name} (AUC = {auc_score:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")

plt.legend()

plt.show()
```

```python
final_model = clone(best_model)

final_model.fit(
    X,
    y
)

print("Final model trained:", best_model_name)
```

```python
df_clean["Previous_Performance_Risk"] = (
    1 -
    (
        df_clean["Previous_Scores"] / 100
    )
)
```

```python
df_clean["Attendance_Risk_Numeric"] = (
    df_clean["Attendance_Risk"]
    .astype(int)
)
```

```python
df_clean["Intervention_Utility"] = (

    0.60 *
    df_clean[
        "Academic_Risk_Probability"
    ]

    +

    0.25 *
    df_clean[
        "Attendance_Risk_Numeric"
    ]

    +

    0.15 *
    df_clean[
        "Previous_Performance_Risk"
    ]
)
```

```python
df_clean["Eligible_For_Intervention"] = (

    (
        df_clean[
            "Academic_Risk_Probability"
        ] >= 0.50
    )

    |

    df_clean[
        "Attendance_Risk"
    ]
)
```

```python
df_clean[
    "Eligible_For_Intervention"
].value_counts()
```

```python
eligible_students = df_clean[
    df_clean[
        "Eligible_For_Intervention"
    ]
].copy()
```

```python
print(
    "Eligible students:",
    len(eligible_students)
)
```

```python
INTERVENTION_CAPACITY = 10
```

```python
random_selection = eligible_students.sample(
    n=min(
        INTERVENTION_CAPACITY,
        len(eligible_students)
    ),
    random_state=42
)
```

```python
random_selection[
    [
        "Academic_Risk_Probability",
        "Attendance_Risk",
        "Intervention_Utility"
    ]
]
```

```python
highest_risk_selection = (
    eligible_students
    .sort_values(
        by="Academic_Risk_Probability",
        ascending=False
    )
    .head(INTERVENTION_CAPACITY)
)
```

```python
highest_risk_selection[
    [
        "Academic_Risk_Probability",
        "Attendance_Risk",
        "Intervention_Utility"
    ]
]
```

```python
from scipy.optimize import milp
from scipy.optimize import LinearConstraint
from scipy.optimize import Bounds
```

```python
optimisation_data = (
    eligible_students
    .reset_index()
    .rename(
        columns={
            "index": "Original_Index"
        }
    )
)
```

```python
utility_values = (
    optimisation_data[
        "Intervention_Utility"
    ].to_numpy()
)
```

```python
n_students = len(
    optimisation_data
)
```

```python
capacity = min(
    INTERVENTION_CAPACITY,
    n_students
)
```

```python
objective = -utility_values
```

```python
integrality = np.ones(
    n_students
)
```

```python
bounds = Bounds(
    lb=np.zeros(n_students),
    ub=np.ones(n_students)
)
```

```python
constraint = LinearConstraint(
    np.ones(
        (
            1,
            n_students
        )
    ),
    lb=capacity,
    ub=capacity
)
```

```python
optimisation_result = milp(
    c=objective,
    integrality=integrality,
    bounds=bounds,
    constraints=constraint
)
```

```python
print(
    optimisation_result.success
)

print(
    optimisation_result.message
)
```

```python
selected_positions = np.where(
    optimisation_result.x > 0.5
)[0]
```

```python
optimised_selection = (
    optimisation_data
    .iloc[selected_positions]
    .copy()
)
```

```python
optimised_selection = (
    optimised_selection
    .sort_values(
        by="Intervention_Utility",
        ascending=False
    )
)
```

```python
optimised_selection[
    [
        "Original_Index",
        "Academic_Risk_Probability",
        "Attendance_Risk",
        "Previous_Scores",
        "Intervention_Utility"
    ]
]
```

```python
def evaluate_strategy(
    selection,
    strategy_name
):

    total_utility = (
        selection[
            "Intervention_Utility"
        ].sum()
    )

    average_risk = (
        selection[
            "Academic_Risk_Probability"
        ].mean()
    )

    attendance_risk_students = (
        selection[
            "Attendance_Risk"
        ].sum()
    )

    high_risk_students = (
        (
            selection[
                "Academic_Risk_Probability"
            ] >= 0.50
        )
        .sum()
    )

    return {
        "Strategy": strategy_name,

        "Students_Selected":
            len(selection),

        "Total_Utility":
            total_utility,

        "Average_ML_Risk":
            average_risk,

        "Attendance_Risk_Students":
            attendance_risk_students,

        "High_ML_Risk_Students":
            high_risk_students
    }
```

```python
strategy_results = []

strategy_results.append(
    evaluate_strategy(
        random_selection,
        "Random Allocation"
    )
)

strategy_results.append(
    evaluate_strategy(
        highest_risk_selection,
        "Highest Risk Allocation"
    )
)

strategy_results.append(
    evaluate_strategy(
        optimised_selection,
        "Optimised Allocation"
    )
)
```

```python
strategy_comparison = pd.DataFrame(
    strategy_results
)

strategy_comparison
```

```python
strategy_comparison.sort_values(
    by="Total_Utility",
    ascending=False
)
```

```python
plt.figure(
    figsize=(8, 5)
)

plt.bar(
    strategy_comparison[
        "Strategy"
    ],
    strategy_comparison[
        "Total_Utility"
    ]
)

plt.xlabel(
    "Intervention Strategy"
)

plt.ylabel(
    "Total Intervention Utility"
)

plt.title(
    "Comparison of Intervention Allocation Strategies"
)

plt.xticks(
    rotation=20
)

plt.show()
```
